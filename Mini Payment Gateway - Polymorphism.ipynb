```python
class Payment:
  def pay(self):
    print("Payment Gateway")
class CreditCard(Payment):
  def pay(self):
    print("Processing Credit Card Payment")
class UPI(Payment):
  def pay(self):
    print("Processing UPI Payment")
class NetBanking(Payment):
  def pay(self):
    print("Processing Net Banking Payment")
Card = CreditCard()
upi = UPI()
Net = NetBanking()
Payment_Gateway = [Card,upi,Net]
for gateway in Payment_Gateway:
  gateway.pay()
```
